```python
import torch
import numpy as np
from PIL import Image
from torch.utils.data import dataset
import os
from torchvision.datasets import ImageFolder
from torchvision import transforms
```

Traditional Method of loading data:

```python
class ImageDataset(dataset.Dataset):
    def __init__(self, image_dir, label,transform=None):
        self.image_dir = image_dir
        self.transform = transform
        self.image_paths = [f for f in os.listdir(image_dir) if f.endswith('.jpg') or f.endswith('.png')]
        self.label = label   
    def __len__(self):
        return len(self.image_paths)
    def __getitem__(self, idx):
        img_name = os.path.join(self.image_dir, self.image_paths[idx])
        image = Image.open(img_name).convert('RGB')
        label = self.label
        if self.transform:
            image = self.transform(image)
        return image, label
        
```

```python
ants_dataset = ImageDataset(
    image_dir='D:/Python Project/Pytorch_practice/hymenoptera/train/ants',label = 'ants')
bees_dataset = ImageDataset(
    image_dir='D:/Python Project/Pytorch_practice/hymenoptera/train/bees',label = 'bees')
train_dataset = ants_dataset + bees_dataset
```

Using Torchvision.datasets.ImageFolder

```python
# Define transformations
transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.ToTensor(),
])

# Load entire dataset (automatically assigns labels)
full_dataset = ImageFolder(
    root='D:/Python Project/Pytorch_practice/hymenoptera/train',
    transform=None
)

# Split into ants and bees datasets (if needed)
ants_indices = [i for i, (_, label) in enumerate(full_dataset) if label == 0]
bees_indices = [i for i, (_, label) in enumerate(full_dataset) if label == 1]

from torch.utils.data import Subset
ants_dataset = Subset(full_dataset, ants_indices)
bees_dataset = Subset(full_dataset, bees_indices)
train_dataset = ants_dataset + bees_dataset
```
